```r
suppressMessages(initial_metadata <- read_csv(initial_metadata_path))
initial_metadata$sampleCollectionDate <- ymd(initial_metadata$sampleCollectionDate)
initial_youngest_date <- max(initial_metadata$sampleCollectionDate, na.rm = TRUE)
suppressMessages(initial_tree_node_ci <- read_csv(initial_tree_node_ci_path))
initial_tree = read.tree(initial_tree_path)
```

```r
initial_tree_plot <- ggtree(initial_tree, mrsd=initial_youngest_date) + 
  		theme_tree2() + # This ggtree theme will add a forward in time x-axis
  		labs(caption="years") 

suppressMessages(
  initial_tree_plot <- initial_tree_plot + geom_tiplab(as_ylab=TRUE)
)
initial_tree_dotted_line_lst <- get_tip_data_and_xend(initial_tree_plot)
initial_tree_plot <- initial_tree_plot +
  geom_segment(
    data = initial_tree_dotted_line_lst$tip_data,
    aes(x = x, y = y, xend = initial_tree_dotted_line_lst$xend, yend = y),
    linetype = "dotted",
    color = "grey60",
    alpha = 0.5
  )

initial_tree_ci <- get_internal_node_ci(
  initial_tree_plot,
  initial_tree_node_ci,
  "interval_0.050000000000000044",
  "upper_0.95"
)
initial_tree_plot <- initial_tree_plot +
  geom_segment(
    data = initial_tree_ci$internal_nodes_ci,
    aes(x = initial_tree_ci$xs,
      xend = initial_tree_ci$xends, 
      y = initial_tree_ci$ys, 
      yend = initial_tree_ci$yends),
    color = 'blue', alpha = 0.2, linewidth = 2
  )
initial_tree_plot
```
