```r
suppressMessages(downsampled_metadata <- read_csv(downsampled_metadata_path))
downsampled_metadata$sampleCollectionDate <- ymd(downsampled_metadata$sampleCollectionDate)
downsampled_youngest_date <- max(downsampled_metadata$sampleCollectionDate, na.rm = TRUE)
suppressMessages(downsampled_node_ci <- read_csv(downsampled_tree_node_ci_path))
```

```r
downsampled_initial_tree = read.tree(downsampled_tree_path)

downsampled_initial_tree_plot <- ggtree(downsampled_initial_tree, mrsd=downsampled_youngest_date) + 
  		theme_tree2() + # This ggtree theme will add a forward in time x-axis
  		labs(caption="years") 
suppressMessages(
  downsampled_initial_tree_plot <- downsampled_initial_tree_plot + geom_tiplab(as_ylab=TRUE)
)
downsampled_initial_tree_dotted_line_lst <- get_tip_data_and_xend(downsampled_initial_tree_plot)
downsampled_initial_tree_plot <- downsampled_initial_tree_plot +
  geom_segment(
    data = downsampled_initial_tree_dotted_line_lst$tip_data,
    aes(x = x, y = y, xend = downsampled_initial_tree_dotted_line_lst$xend, yend = y),
    linetype = "dotted",
    color = "grey60",
    alpha = 0.5
  )

downsampled_initial_tree_ci <- get_internal_node_ci(
  downsampled_initial_tree_plot,
  downsampled_node_ci,
  "interval_0.050000000000000044",
  "upper_0.95"
)
downsampled_initial_tree_plot <- downsampled_initial_tree_plot +
  geom_segment(
    data = downsampled_initial_tree_ci$internal_nodes_ci,
    aes(x = downsampled_initial_tree_ci$xs,
      xend = downsampled_initial_tree_ci$xends, 
      y = downsampled_initial_tree_ci$ys, 
      yend = downsampled_initial_tree_ci$yends),
    color = 'blue', alpha = 0.2, linewidth = 2
  )
downsampled_initial_tree_plot
```
