```r
options(warn = -1) # Suppress warnings for cleaner output
#Load libraries
suppressPackageStartupMessages({
    library(treeio)
    library(ggtree)
    library(ggplot2)
    library(readr)
    library(lubridate)
    library(dplyr)
    library(png)
    library(grid)
    library(tidyr)
    library(htmltools)
}
)
```

```r
# Function for producing CI.95 from plots and dataframe of CIs.
get_internal_node_ci <- function(tree_plot, node_ci, lower_col, upper_col) {
  plot_data <- tree_plot$data
  merged_plot_ci_data <- node_ci %>%
    left_join(plot_data, by = c("node" = "label"))
  internal_nodes_ci <- merged_plot_ci_data %>% filter(isTip == FALSE)
  xs <- internal_nodes_ci[[lower_col]]
  xends <- internal_nodes_ci[[upper_col]]
  ys <- internal_nodes_ci$y
  yends <- internal_nodes_ci$y
  list(xs = xs, xends = xends, ys = ys, yends = yends, internal_nodes_ci = internal_nodes_ci)
}
# Function for plotting getting max x value from tree plot data.
get_tip_data_and_xend <- function(tree_plot) {
  tip_data <- tree_plot$data %>% dplyr::filter(isTip)
  xend <- max(ggplot_build(tree_plot)$layout$panel_params[[1]]$x.range) + 1
  list(tip_data = tip_data, xend = xend)
}
```
