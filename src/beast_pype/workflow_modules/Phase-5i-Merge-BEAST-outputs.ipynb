# Merging Log & Tree Files.

```bash
log_file_burnin=''
log_file_output=''
log_files_to_combine=''
trees_file_burnin=''
trees_file_output=''
trees_files_to_combine=''
kernel_name='beast_pype'
```

```bash
# For some reason this still has to be called so that slurm's sbatch to use the beast_pype environment.
source activate $kernel_name
```

## Merge Logfiles

```bash
logcombiner -b $log_file_burnin -log $log_files_to_combine  -o $log_file_output
```

## Merge Treefiles

```bash
logcombiner -b $trees_file_burnin -log $trees_files_to_combine  -o $trees_file_output
```
